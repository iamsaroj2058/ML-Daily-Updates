#STEP 1. Extract input and outputs cols

### now  we  are extracting  dependent and independent variables    [CGPA and  iq is independent variable and  placement is dependebnt variable because  placement is  depends  on both cgpa and iq]



*   taking both cgps and iq in X which is independent variable
*   taking placement in y which is dependent variable






```python
X = df.iloc[:,0:2]
y = df.iloc[:,-1]
```

```python
X
```

```python
y
```

# 2.Train Test Split

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size = 0.1)
```

```python
X_train
```

```python
y_train
```

```python
X_test
```

```python
y_test
```

###  scale the data   not necessary but good to go with

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train =scaler.fit_transform(X_train)
```

```python
X_train
```

### what  we  do here is we have scale the data from -1 to 1  now same for other

```python
X_test = scaler.transform(X_test)
```

```python
X_test
```

# 4. MOdel **training**

```python
#  i have desided to use logistic regression classifier so
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf = LogisticRegression()
```

```python
#this is model  rtraining
clf.fit(X_train,y_train)
```

# Evaluate the model

```python
#what we  exactly do here is evaluate the  accuracy from the test data set
```

```python
y_pred =clf.predict(X_test)
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```

```python
#how to plot the decision boundry(decision boundry mean --> what pattern does  ml model recognize to predict the result)


from mlxtend.plotting import plot_decision_regions

fig = plot_decision_regions(X=X_train, y=y_train.values,
                                clf=clf, legend=2)

```
